```python
raw_export = raw_export.splitlines()
raw_export
```

```python
human_line_number = []
blank_line_number = []
header_line_number = []
skipped_lines = []

for line_no, line in enumerate(raw_export, start=1):
    human_line_number.append(line_no)

    if line.startswith('#') or line.lower().startswith('date'):
        header_line_number.append(line_no)
        skipped_lines.append(line_no)

    elif not line.strip():
        blank_line_number.append(line_no)
        skipped_lines.append(line_no)
    
    else:
        continue


print(human_line_number)
print(blank_line_number)
print(header_line_number)
print(skipped_lines)
```
